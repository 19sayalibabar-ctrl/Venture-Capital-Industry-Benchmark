```python
# Group by industry
industry_summary = df.groupby('industry').agg({
    'funding_amount_usd': 'mean',
    'estimated_revenue_usd': 'mean',
    'estimated_valuation_usd': 'mean',
    'employee_count': 'mean',
    'startup_age': 'mean'
}).reset_index()

# Round for better display
industry_summary = industry_summary.round(2)

# Rename columns for clarity
industry_summary.columns = ['Industry', 'Avg Funding (USD)', 'Avg Revenue (USD)',
                            'Avg Valuation (USD)', 'Avg Employee Count', 'Avg Startup Age']
```

```python
industry_summary['Valuation-to-Revenue'] = industry_summary['Avg Valuation (USD)'] / industry_summary['Avg Revenue (USD)']
industry_summary['Funding-to-Revenue'] = industry_summary['Avg Funding (USD)'] / industry_summary['Avg Revenue (USD)']
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 6))
sns.barplot(x='Valuation-to-Revenue', y='Industry', data=industry_summary.sort_values('Valuation-to-Revenue', ascending=False))
plt.title('Valuation-to-Revenue Ratio by Industry')
plt.xlabel('Valuation / Revenue')
plt.ylabel('Industry')
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 6))
barplot = sns.barplot(
    x='Avg Valuation (USD)', 
    y='Industry', 
    data=industry_summary.sort_values('Avg Valuation (USD)', ascending=False),
    palette='Blues_d'
)

# Add values at the end of bars
for index, row in industry_summary.iterrows():
    barplot.text(row['Avg Valuation (USD)'] + 1e7, index, f"{row['Avg Valuation (USD)']/1e9:.2f}B", va='center')

plt.title('Average Valuation by Industry')
plt.xlabel('Avg Valuation (USD)')
plt.tight_layout()
plt.show()
```

```python

Xx = df.drop(columns=['exited'])  
Yy = df['exited']
```

```python
scaler = MinMaxScaler()
Xx_scaled = scaler.fit_transform(Xx)
```

```python
Xx_scaled_df = pd.DataFrame(Xx_scaled, columns=Xx.columns)
Xx_scaled_df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    Xx_scaled_df, Yy, test_size=0.2, random_state=42, stratify=Yy 
)
```

```python
print(y_train.value_counts())
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

print("After resampling:")
print(y_train_res.value_counts())
```

```python
model.fit(X_train_res, y_train_res)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, f1_score

# Step 1: Train the model on the resampled data
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train_res, y_train_res)

# Step 2: Predict on original (imbalanced) test set
y_pred = model.predict(X_test)

# Step 3: Evaluation
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred, zero_division=0))

# Optional: just F1-score
f1 = f1_score(y_test, y_pred)
print("\nF1 Score:", f1)
```
